```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import pandas as pd
```

```python
plt.rcParams.update({
    "text.usetex": False, # Use True apenas se tiver uma distribuição LaTeX instalada no PC
    "font.family": "serif",
    "mathtext.fontset": "cm", # Usa a fonte Computer Modern (padrão do LaTeX)
    "axes.labelsize": 12
})
```

```python
d1 = pd.read_csv('100uf21k.csv')
d2 = pd.read_csv('100uf100k.csv')
d3 = pd.read_csv('10uf100k.csv')
d4 = pd.read_csv('10uf21k.csv')
```

```python
array = []
array.append(d1)
array.append(d2)
array.append(d3)
array.append(d4)
```

```python
for i in range(len(array)):
    array[i]['t'] = array[i]['t']/1000
```

```python
def vt(t, V0, rc):
    e = np.exp(-t/rc)
    return V0*(1-e)
```

```python
titulos = [r'Tensão no capacitor - $C = 93,3 \ \mu F$, $R = 21,92 \ k\Omega$', r'Tensão no capacitor - $C = 93,3 \ \mu F, R = 81,6 \ k\Omega$', 
           r'Tensão no capacitor - $C = 11,8 \ \mu F, R = 81,6 \ k\Omega$', r'Tensão no capacitor - $C = 11,8 \ \mu F, R = 21,92 \ k\Omega$']
p = []
```

```python
for i in range(len(array)):
    param, _ = curve_fit(vt,array[i]['t'], array[i]['v'], p0=[5,2])
    label_texto = (
    f"Ajuste $V(t) = V_0 (1-e^{{-t/\\tau}})$\n"
    f"$V_0$ = {param[0]:.3f}, $\\tau$ = {param[1]:.3f}"
    )
    p.append(param)
    plt.scatter(array[i]['t'], array[i]['v'], label='Dados Experimentais', s=10, c='black', marker='o', edgecolors='none')
    plt.plot(array[i]['t'], vt(array[i]['t'], *param), 'r-', label=label_texto)
    plt.yticks([0,1,2,3,4,5])
    plt.title(titulos[i])
    plt.grid(True)
    plt.xlabel('t (s)')
    plt.ylabel('V(t) (V)')
    plt.legend(loc='lower right')
    nome_seguro = titulos[i].replace(" ", "_")
    for carac in r'\$/{}:^': # Lista de caracteres problemáticos
        nome_seguro = nome_seguro.replace(carac, "")
    # Agora salvamos com o nome limpo
    plt.savefig(f'{nome_seguro}.png')
    plt.close()
    plt.show()
```

```python
def erro(qe,qt):
    return np.abs(qe-qt)/qt * 100
```

```python
tensao = np.array([
    4.11, 4.21, 4.09, 4.17, 4.22, 2.50, 2.51, 2.64, 2.57, 2.47,
    2.59, 4.18, 4.08, 4.21, 4.01, 3.70, 3.70, 3.75, 3.51, 2.93,
    3.01, 3.73, 2.35, 2.26, 4.14, 4.20, 4.26, 4.13, 4.23
])
```

```python
medida = range(1, len(tensao)+1)
```

```python
plt.scatter(medida, tensao,s=10, c='black', marker='o', edgecolors='none')
plt.xlabel("Medida")
plt.ylabel("Tensão no resistor (V)")
plt.grid(True)
```

```python
fig, ax = plt.subplots(figsize=(9, 5.5))

# --- 3. Plotando os pontos (Scatter) ---
ax.scatter(medida, tensao, s=10, c='black', marker='o', edgecolors='none', label='Medidas', zorder=3)

# --- 4. Adicionando as Regiões Sombreadas Verticais ---
# Capturamos os limites atuais do eixo Y para que o sombreado cubra toda a altura do gráfico
ymin, ymax = ax.get_ylim()

# O 'fill_between' vai preencher de 'ymin' até 'ymax' na vertical, 
# mas APENAS nos pontos do eixo X onde a condição (tensao < 3.75) for verdadeira.
ax.fill_between(medida, ymin, ymax, where=(tensao < 3.8), 
                color='#e74c3c', alpha=0.15, zorder=1, label='Obstrução da passagem de luz')

# Reaplicamos os limites originais do eixo Y para que o preenchimento não altere o zoom do gráfico
ax.set_ylim(ymin, ymax)

# --- 5. Customização de Títulos e Eixos ---
ax.set_xlabel("Medida", fontsize=11, labelpad=10)
ax.set_ylabel("Tensão no resistor (V)", fontsize=11, labelpad=10)
ax.set_title("Análise de Tensão no Resistor por Medida", fontsize=14)

# --- 6. Estilização Estética ---
ax.grid(True, linestyle=':', alpha=0.6, zorder=2) # Grade pontilhada sutil atrás dos pontos
ax.spines['top'].set_visible(False)              # Remove a linha de cima da borda
ax.spines['right'].set_visible(False)            # Remove a linha da direita da borda
ax.spines['left'].set_color('#cccccc')           # Suaviza a cor da borda esquerda
ax.spines['bottom'].set_color('#cccccc')         # Suaviza a cor da borda inferior

# --- 7. Legenda ---
ax.legend(loc='lower right', frameon=True, facecolor='white', edgecolor='#e2e8f0', framealpha=0.9, fontsize=10)

plt.tight_layout()
plt.savefig(f'tensao ldr.png')
plt.show()
```
