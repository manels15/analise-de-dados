```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import pandas as pd
```

```python
plt.rcParams.update({
    "text.usetex": False, # Use True apenas se tiver uma distribuição LaTeX instalada no PC
    "font.family": "serif",
    "mathtext.fontset": "cm", # Usa a fonte Computer Modern (padrão do LaTeX)
    "axes.labelsize": 12
})
```

```python
d1 = pd.read_csv('100uf21k.csv')
d2 = pd.read_csv('100uf100k.csv')
d3 = pd.read_csv('10uf100k.csv')
d4 = pd.read_csv('10uf21k.csv')
```

```python
array = []
array.append(d1)
array.append(d2)
array.append(d3)
array.append(d4)
```

```python
for i in range(len(array)):
    array[i]['t'] = array[i]['t']/1000
```

```python
def vt(t, V0, rc):
    e = np.exp(-t/rc)
    return V0*(1-e)
```

```python
titulos = [r'Tensão no resistor - $C = 93,3 \ \mu F$, $R = 21,92 \ k\Omega$', r'Tensão no resistor - $C = 93,3 \ \mu F, R = 9,78 \ k\Omega$', 
           r'Tensão no resistor - $C = 11,8 \ \mu F, R = 9,78 \ k\Omega$', r'Tensão no resistor - $C = 11,8 \ \mu F, R = 21,92 \ k\Omega$']
p = []
```

```python
for i in range(len(array)):
    param, _ = curve_fit(vt,array[i]['t'], array[i]['v'], p0=[5,2])
    label_texto = (
    f"Ajuste $V(t) = V_0 (1-e^{{-t/\\tau}})$\n"
    f"$V_0$ = {param[0]:.3f}, $\\tau$ = {param[1]:.3f}"
    )
    p.append(param)
    plt.scatter(array[i]['t'], array[i]['v'], label='Dados Experimentais')
    plt.plot(array[i]['t'], vt(array[i]['t'], *param), 'r-', label=label_texto)
    plt.yticks([0,1,2,3,4,5])
    plt.title(titulos[i])
    plt.grid(True)
    plt.xlabel('t (s)')
    plt.ylabel('V(t) (V)')
    plt.legend(loc='lower right')
    nome_seguro = titulos[i].replace(" ", "_")
    for carac in r'\$/{}:^': # Lista de caracteres problemáticos
        nome_seguro = nome_seguro.replace(carac, "")
    # Agora salvamos com o nome limpo
    plt.savefig(f'{nome_seguro}.png')
    plt.close()
    plt.show()
```

```python
def erro(qe,qt):
    return np.abs(qe-qt)/qt * 100
```
